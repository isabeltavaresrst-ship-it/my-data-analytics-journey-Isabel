```python
# Part 2: Data Cleaning

# 1. Handling Missing Values

# Create copies of original DataFrames for cleaning

customers_clean = customers.copy()
products_clean = products.copy()
stores_clean = stores.copy()
sales_clean = sales.copy()
inventory_clean = inventory.copy()
```

```python
# Identify all missing values in each dataset before cleaning

dataframes = {
    "customers_clean": customers_clean,
    "products_clean": products_clean,
    "stores_clean": stores_clean,
    "sales_clean": sales_clean,
    "inventory_clean": inventory_clean
}

for name, df in dataframes.items():
    print(name.upper())
    display(df.isnull().sum())
```

```python
# For date columns with missing values, use forward fill or backward fill

date_columns = {
    "customers_clean": ["registration_date"],
    "stores_clean": ["opening_date"],
    "sales_clean": ["date"],
    "inventory_clean": ["last_restock_date"]
}

for df_name, columns in date_columns.items():
    df = dataframes[df_name]
    for col in columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")
        df[col] = df[col].ffill()
        df[col] = df[col].bfill()
```

```python
# For numerical columns with missing values, replace with column mean
for name, df in dataframes.items():
    numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns
    
    for col in numeric_columns:
        if df[col].isnull().sum() > 0:
            df[col] = df[col].fillna(df[col].mean())
```

```python
# For categorical columns with missing values, replace them with the most frequent value

for name, df in dataframes.items():

    categorical_columns = df.select_dtypes(include=["str"]).columns

    for col in categorical_columns:
        if df[col].isnull().sum() > 0:
            df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Check missing values after cleaning

for name, df in dataframes.items():
    print(name.upper())
    display(df.isnull().sum())
```

```python
# Check for duplicate entries in customers and products

print("Customer duplicates before cleaning:", customers_clean.duplicated().sum())
print("Product duplicates before cleaning:", products_clean.duplicated().sum())

customers_clean = customers_clean.drop_duplicates()
products_clean = products_clean.drop_duplicates()

print("Customer duplicates after cleaning:", customers_clean.duplicated().sum())
print("Product duplicates after cleaning:", products_clean.duplicated().sum())
```

### Duplicate Removal Explanation

I checked for duplicate records in the customers and products dataframes using the `duplicated()` function. This function identifies rows where all values are repeated. After checking the number of duplicate rows, I used `drop_duplicates()` to remove any repeated records. This approach helps prevent duplicate customers or products from affecting later analysis.

## What are the key advantages of using Pandas for data cleaning compared to other methods?
                                            
Pandas makes data cleaning faster and more efficient because it provides built-in functions for handling missing values, removing duplicates, changing data types, and filtering data. Instead of manually editing rows in Excel, Pandas allows large datasets to be cleaned using a few lines of code. ## How would your approach to handling missing values differ if the missing data was not random but had a pattern or meaning?

If the missing data followed a pattern, I would first investigate why the values were missing before replacing them. For example, if customer ages were missing only for certain regions or stores, simply filling them with the average could create misleading results. In that situation, I would analyze the cause of the missing values, create a separate category for missing data if appropriate, or use a more targeted method of imputation based on related information.



## What types of data quality issues might not be immediately visible through simple DataFrame inspection methods?

Some data quality issues are not obvious when viewing the first few rows of a DataFrame. Examples include inconsistent spellings, duplicate records with slight differences, incorrect data entry, outliers, and values stored in the wrong format. For example, a state name could appear as both "CA" and "California", which may not be immediately visible but could affect analysis results. These issues often require additional checks and validation beyond basic inspection.

## In what scenarios might it be better to remove rows with missing values rather than imputing them?

Removing rows may be a better option when only a small number of records contain missing values and those records are not important to the overall analysis. It may also be appropriate when critical information is missing and there is no reliable way to estimate it. For example, if a sales transaction is missing both the product and price information, filling in values could introduce errors, so removing that record may produce more accurate results.
