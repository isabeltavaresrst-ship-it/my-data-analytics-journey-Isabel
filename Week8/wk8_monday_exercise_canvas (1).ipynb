# 1. Load a CSV file

```python
# 2. Changing the Delimiter (sep)
# If your file uses a semicolon (;) or tab (\t) instead of a comma.
# Assuming the file used semicolons
df = pd.read_csv('students.csv', sep=',')
```

```python
df
```

```python
# 3. Setting a Column as the Index (index_col)
# If you want to use a specific column (like Name) as the row identifier instead of the default numeric index.
df = pd.read_csv('students.csv', index_col ='Name')
```

```python
df
```

```python
# 4. Selecting Specific Columns (usecols)
# Useful when you have a massive dataset and only need a few specific columns, which saves memory.
df = pd.read_csv('students.csv',usecols=['Name', 'Grade'])
```

```python
df
```

```python
# 5. Renaming Columns on Import (names)
# Use this to overwrite the headers or if the file has no header row. 
# You must set header=0 (to skip the original) if you want to replace it.
df = pd.read_csv('students.csv', names=['Student', 'Years', 'Score', 'Location'], header =0)
```

```python
df
```

```python
# 6. Defining Data Types (dtype)
# You can force a column to be a specific type (e.g., Age as a float or Name as a category) to optimize performance.
df = pd.read_csv('students.csv', dtype={'Age':'float', 'Grade':'float'})
```

```python
df
```

```python
df.info()
```

```python
# 7. Limiting the Number of Rows (nrows)
# Best for testing code on large datasets without reading the entire file.
# Read only the first 5 rows
df = pd.read_csv('students.csv',nrows=5)
```

```python
df
```

```python
# 8. Skipping Rows (skiprows)
# Useful if your CSV file has descriptive text or metadata at the top that isn't part of the actual data.
# Skip the first 2 lines of the file
df = pd.read_csv('students.csv', skiprows=2)
```

```python
df
```

```python
df = pd.read_csv('students.csv')
```

```python
cols_names = df.columns
```

```python
df = pd.read_csv('students.csv', skiprows=3, names=cols_names)
```

```python
df
```

```python
# 9. Handling Missing Values (na_values)
# If your data represents "empty" or "unknown" entries with specific strings like "N/A" or "None", 
# you can tell pandas to treat them as NaN (Not a Number).
df = pd.read_csv('students_updated.csv', na_values=['N/A', 'unknowkn'])
```

```python
df
```

```python
# 10. Parsing Dates (parse_dates)
# If you had a date column (e.g., 'EnrollmentDate'), this would automatically convert those strings 
# into actual datetime objects.
df = pd.read_csv('students_updated.csv', parse_dates=['EnrollmentDate'])
```

```python
df
```

```python
# 2. Check for nulls
df.isnull().sum()
```

```python
# 3. Handle nulls (e.g., fill with the mean)
df['Age'] = df ['Age'].fillna(df['Age'].mean())
```

```python
df
```

```python
from sqlalchemy import create_engine
```

```python
# 1. Create the engine
engine = create_engine('sqlite:///bike_orders_database.sqlite')
```

```python
# 2. Read the data directly using the engine
# You can pass the engine directly to read_sql
df = pd.read_sql('SELECT * FROM bikes', engine, index_col='index')
```
