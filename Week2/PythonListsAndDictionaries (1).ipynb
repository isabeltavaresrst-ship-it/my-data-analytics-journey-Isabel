```python
#1
shopping_list = ["milk", "eggs", "strawberries", "bread", "jam", "cheese"]
```

```python
#5
shopping_list_prices = {"eggs": 5.99, "bread": 1.88, "jam": 3.99}
print(shopping_list_prices)
```
